```python
from rapidfuzz import fuzz

def fuzzy_match(row, candidates, threshold=90):
    best_match = None
    best_score = 0

    for idx in candidates:
        score = fuzz.token_set_ratio(
            row['Work_description_clean'],
            fuzzy_data.loc[idx, 'Work_description_clean']
        )

        if score >= threshold and score > best_score:
            best_score = score
            best_match = idx

    return best_match, best_score
```

```python
from rapidfuzz import process, fuzz

# Get unique project descriptions
unique_projects = project_summary[
    ['Project_ID', 'Constituency', 'Work_Description', 'Sanction_Amount']
].copy()

print("Unique projects:", len(unique_projects))
```

```python
unique_projects['Work_Description'] = (
    unique_projects['Work_Description']
    .astype(str)
    .str.lower()
    .str.strip()
)
```

```python
potential_matches = []

for constituency, group in unique_projects.groupby('Constituency'):
    indices = group.index.tolist()

    descriptions = group['Work_Description'].to_dict()

    for idx in indices:
        current = descriptions[idx]

        matches = process.extract(
            current,
            descriptions,
            scorer=fuzz.ratio,
            score_cutoff=95,
            limit=5
        )

        for match_text, score, match_idx in matches:
            if idx != match_idx:
                potential_matches.append({
                    'Project_ID_1': group.loc[idx, 'Project_ID'],
                    'Project_ID_2': group.loc[match_idx, 'Project_ID'],
                    'Description_1': current,
                    'Description_2': match_text,
                    'Similarity': score,
                    'Amount_1': group.loc[idx, 'Sanction_Amount'],
                    'Amount_2': group.loc[match_idx, 'Sanction_Amount']
                })

potential_matches_df = pd.DataFrame(potential_matches)

print("Potential matches:", len(potential_matches_df))
```

```python
potential_matches_df['Amount_1'] = pd.to_numeric(
    potential_matches_df['Amount_1'],
    errors='coerce'
)

potential_matches_df['Amount_2'] = pd.to_numeric(
    potential_matches_df['Amount_2'],
    errors='coerce'
)
```

```python

potential_matches_df['Amount_Difference_%'] = (
    abs(
        potential_matches_df['Amount_1'] -
        potential_matches_df['Amount_2']
    )
    / potential_matches_df[['Amount_1', 'Amount_2']].max(axis=1)
) * 100
```

```python
strong_candidates = potential_matches_df[
    potential_matches_df['Amount_Difference_%'] <= 2
].copy()

print("Strong candidates:", len(strong_candidates))
```
